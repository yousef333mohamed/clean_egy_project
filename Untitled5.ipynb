### [ **Workforce** **data** & **Workforce** **attendance** **data**](https://)

### **Purpose** **of** **Workforce** **Data**

`workforce.csv` is a **worker-level master table**: one row per collection-crew worker, with
their governorate, region, assigned shift, and years of experience. It describes **who**
operates the system.

### **Purpose** **of** **Attendance** **Data**

`workforce_attendance.csv` is a **worker-day panel**: one row per worker per day, recording
whether the worker was present, how many tasks they completed, how much overtime they worked,
and their performance score for that day. It describes **how the workforce actually performed**
day to day.

### **Why** **These** **Datasets** **are** **important**

In this section, we analyze how worker characteristics, such as experience, shift, and region, are related to attendance and daily performance. The goal is to understand whether these factors affect how workers perform their daily tasks.

### [**Environmental** **Data** & **Operational** **Data**](https://)

### Purpose of Environmental Data

The environmental dataset captures daily contextual conditions per region: temperature,
rainfall, traffic level, and calendar flags (holiday, festival, weekend).These variables represent the operational performance of the waste management system

### Purpose of Operational Data

The operational dataset captures daily bin-level service outcomes: whether a collection
was missed, whether an emergency request was raised, whether a complaint was filed, whether
illegal dumping was flagged, and how long the service took to complete. These are the direct
performance indicators of the waste collection system.

### Why These Datasets Are Important

Together, they allow us to ask: does the environment in which the system operates affect how
well it performs? Understanding these relationships provides valuable insights that can support future predictive analytics and operational decision-making.

##  Data Dictionary

### 1. `environmental_daily.csv`

| Column | Meaning | Data Type | Possible Values | Why It Matters |
|---|---|---|---|---|
| `date` | Calendar date of the observation | Date (string, `YYYY-MM-DD`) | 2026-01-01 → 2026-03-31 | Anchors every environmental reading to a specific day; needed to join conceptually with operational data. |
| `region` | Egyptian macro-region | Categorical | Greater Cairo, Delta, Canal & Sinai, Upper Egypt, Coastal & Red Sea | Determines geographic context and allows regional comparison of conditions and outcomes. |
| `temperature_c` | Average daily temperature | Numeric (float, °C) | ~4.7 to ~33.8 | High/low temperature can affect waste decomposition speed, odor, and crew productivity. |
| `rainfall_mm` | Daily rainfall amount | Numeric (float, mm) | 0 to ~5.2 (mostly 0) | Rain can slow trucks, delay collection, and increase missed collections. |
| `is_holiday` | Whether the day is an official holiday | Boolean | True / False | Holidays can change waste generation patterns and staffing levels. |
| `is_festival` | Whether the day is a festival day | Boolean | True / False | Festivals typically increase waste volume and public activity. |
| `is_weekend` | Whether the day falls on a weekend | Boolean | True / False | Weekend staffing and traffic patterns differ from weekdays. |
| `traffic_level` | Categorical daily traffic intensity | Categorical | Low, Medium, High | Traffic directly affects truck routing time and service completion time. |

## Weekly operational patterns

```python
# Create day of week
op_df['day_of_week'] = op_df['date'].dt.day_name()

event_cols = [
    'missed_collection',
    'emergency_request',
    'complaint_filed',
    'illegal_dumping_flag'
]

weekly_ops = (
    op_df.groupby('day_of_week')[event_cols]
    .mean()
    .reindex(dow_order)
    * 100
)

fig, ax = plt.subplots(figsize=(12,6))

weekly_ops.plot(
    kind='bar',
    ax=ax,
    color=[
        PRIMARY_COLOR,
        ACCENT_COLOR,
        "#6AA84F",
        "#C0504D"
    ]
)

ax.set_title("Operational Event Rates by Day of Week")
ax.set_xlabel("Day of Week")
ax.set_ylabel("Occurrence Rate (%)")

plt.xticks(rotation=30)

plt.legend(
    title="Event",
    bbox_to_anchor=(1.02,1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

Both environmental conditions and operational event rates remain relatively stable across the days of the week. While small fluctuations are observed, no clear weekly pattern emerges. This suggests that the day of the week alone is unlikely to be a major factor influencing waste collection performance during the study period.

### **Outlier** **Analysis**


```python
def iqr_outlier_bounds(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

numeric_vars = {
    'temperature_c': env_df['temperature_c'],
    'rainfall_mm': env_df['rainfall_mm'],
    'service_completion_time_min': op_df['service_completion_time_min'].dropna()
}

outlier_summary = []
for name, series in numeric_vars.items():
    lower, upper = iqr_outlier_bounds(series)
    outliers = series[(series < lower) | (series > upper)]
    outlier_summary.append({
        'variable': name,
        'lower_bound': round(lower, 2),
        'upper_bound': round(upper, 2),
        'n_outliers': len(outliers),
        'pct_outliers': round(len(outliers) / len(series) * 100, 2)
    })

pd.DataFrame(outlier_summary)
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(x=env_df['rainfall_mm'], color=ACCENT_COLOR)
plt.title('Rainfall Outliers Are Simply the Rainy Days')
plt.xlabel('Rainfall (mm)')
plt.tight_layout()
plt.show()
```

 No outliers were removed. The rainfall "outliers" are legitimate,
climate-consistent rain events and are analytically important (not noise) — removing them

### **Outlier** **Analysis**


```python
def iqr_outlier_bounds(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

numeric_vars = {
    'temperature_c': env_df['temperature_c'],
    'rainfall_mm': env_df['rainfall_mm'],
    'service_completion_time_min': op_df['service_completion_time_min'].dropna()
}

outlier_summary = []
for name, series in numeric_vars.items():
    lower, upper = iqr_outlier_bounds(series)
    outliers = series[(series < lower) | (series > upper)]
    outlier_summary.append({
        'variable': name,
        'lower_bound': round(lower, 2),
        'upper_bound': round(upper, 2),
        'n_outliers': len(outliers),
        'pct_outliers': round(len(outliers) / len(series) * 100, 2)
    })

pd.DataFrame(outlier_summary)
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(x=env_df['rainfall_mm'], color=ACCENT_COLOR)
plt.title('Rainfall Outliers Are Simply the Rainy Days')
plt.xlabel('Rainfall (mm)')
plt.tight_layout()
plt.show()
```

 No outliers were removed. The rainfall "outliers" are legitimate,
climate-consistent rain events and are analytically important (not noise) — removing them

# [**Overall** **System** **Analysis**](https://)


Each dataset was already explored on its own. Here we only look at relationships **across**
datasets — how workforce, bins, trucks, and environment relate to each other at the region and
day level. Cleaning, missing values, and distributions are not repeated here.

```python
regions = sorted(env_df['region'].unique())
print("Regions used across all datasets:", regions)

for name, d in [("operational", op_df), ("smart_bins", bins_df), ("workforce", workforce_df), ("trips", trips)]:
    print(f"{name} regions match environmental: {sorted(d['region'].unique()) == regions}")
```

All five datasets that carry a `region` column use the exact same five regions, so region can be used as a reliable key to connect every part of the system.

## Building Daily System-Level Tables

To connect workforce, bins, trucks, and operations, each dataset is first collapsed to one row
per **region per day**.

```python
# Attendance -> daily region-level workforce summary
attendance_wf = attendance_df.merge(workforce_df[['worker_id', 'region']], on='worker_id', how='left')
daily_workforce = attendance_wf.groupby(['date', 'region']).agg(
    attendance_rate=('present', 'mean'),
    avg_performance=('performance_score', 'mean'),
    avg_completed_tasks=('completed_tasks', 'mean')
).reset_index()
daily_workforce['attendance_rate'] *= 100

# Operational -> daily region-level service summary
daily_ops = op_df.groupby(['date', 'region']).agg(
    missed_collection_rate=('missed_collection', 'mean'),
    complaint_rate=('complaint_filed', 'mean'),
    emergency_rate=('emergency_request', 'mean'),
    avg_completion_time=('service_completion_time_min', 'mean')
).reset_index()
for c in ['missed_collection_rate', 'complaint_rate', 'emergency_rate']:
    daily_ops[c] *= 100

# Smart bin readings -> daily region-level fill summary
readings_region = readings_df.merge(bins_df[['bin_id', 'region']], on='bin_id', how='left')
readings_region['date'] = readings_region['timestamp'].dt.floor('D')
daily_bins = readings_region.groupby(['date', 'region']).agg(
    avg_fill_level=('fill_level_pct', 'mean'),
    avg_waste_weight=('waste_weight_kg', 'mean')
).reset_index()

# Truck trips -> daily region-level truck summary
daily_trucks = trips.groupby(['date', 'region']).agg(
    trips_completed=('status', lambda s: (s == 'Completed').sum()),
    total_trips=('status', 'count'),
    avg_trip_duration=('trip_duration_min', 'mean'),
    total_load_kg=('load_kg', 'sum')
).reset_index()

print(daily_workforce.shape, daily_ops.shape, daily_bins.shape, daily_trucks.shape)
```

```python
system_daily = (
    daily_ops
    .merge(daily_workforce, on=['date', 'region'], how='left')
    .merge(daily_bins, on=['date', 'region'], how='left')
    .merge(daily_trucks, on=['date', 'region'], how='left')
    .merge(env_df, on=['date', 'region'], how='left')
)

print("Combined daily system table:", system_daily.shape)
system_daily.head()
```

`system_daily` now has one row per region per day, combining service outcomes, workforce attendance, bin fill levels, truck activity, and environmental conditions in a single table.

##  Final System-Level Insights

- Region is a reliable common key: all five region-tagged datasets share the exact same five regions.
- Workforce attendance and performance show little connection to missed collections or completion time — service outcomes are mostly independent of daily crew presence.
- Bin fill level does not clearly drive truck trip volume or emergency requests at the region or day level.
- Traffic and rainfall have only a small effect on truck trip duration and fuel efficiency.
- Attendance, fill level, and truck trips each follow their own daily pattern and don't move together over the 90-day period.
- The cross-dataset correlation matrix confirms that, overall, the four systems (workforce, bins, trucks, environment) operate largely independently, with missed collections and complaints being the one strong link found in the data.
